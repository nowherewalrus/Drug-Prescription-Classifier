# 🧪 Drug Prescription Classification using Decision Tree

## 📌 Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn import metrics
```

## 📂 Load and Inspect Dataset

```python
# Load dataset
df = pd.read_csv('drug200.csv')

# Display first few rows
df.head()
```

## 🔄 Data Preprocessing
Encode Categorical Variables

```python
# Encode gender
df = df.replace({'F': 0, 'M': 1})

# Encode blood pressure and cholesterol levels
df = df.replace({'LOW': 0, 'NORMAL': 1, 'HIGH': 2})
```

## 🎯 Feature Selection and Target Variable

```python
# Feature matrix
x = df[['Age', 'Sex', 'BP', 'Cholesterol', 'Na_to_K']]

# Target vector
y = df['Drug']
```

## 🔀 Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=4
)
```

## 🌳 Train Decision Tree Classifier

```python
# Initialize Decision Tree model
drug_tree = DecisionTreeClassifier(
    criterion="entropy",
    max_depth=4,
    random_state=4
)

# Train the model
drug_tree.fit(X_train, y_train)
```

## 📊 Model Evaluation

```python
# Predict on test data
y_pred = drug_tree.predict(X_test)

# Evaluate accuracy
accuracy = metrics.accuracy_score(y_test, y_pred)
print(f"Decision Tree Accuracy: {accuracy:.4f}")

```

## 🌲 Visualize the Decision Tree

```python
plt.figure(figsize=(15, 10))
plot_tree(
    drug_tree,
    filled=True,
    feature_names=['Age', 'Sex', 'BP', 'Cholesterol', 'Na_to_K'],
    class_names=drug_tree.classes_
)
plt.title("Decision Tree for Drug Prescription")
plt.show()
```

## ✅ Summary

✔ Clean preprocessing with categorical encoding
